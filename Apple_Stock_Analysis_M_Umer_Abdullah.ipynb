## Step 6 — Algorithm Analysis (DAA Core)

```python
# Bubble Sort — O(n²)
def bubble_sort(arr):
    arr = arr.copy(); n = len(arr)
    for i in range(n):
        for j in range(n - i - 1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
    return arr

sample = df['Close'].tail(100).tolist()
t0 = time.time(); bubble_sort(sample); t1 = time.time()
print(f'Bubble Sort (n=100) → {(t1-t0)*1000:.4f} ms | Complexity: O(n²)')
```

```python
# Merge Sort — O(n log n)
def merge_sort(arr):
    if len(arr) <= 1: return arr
    mid = len(arr)//2
    L, R = merge_sort(arr[:mid]), merge_sort(arr[mid:])
    res, i, j = [], 0, 0
    while i < len(L) and j < len(R):
        if L[i] <= R[j]: res.append(L[i]); i += 1
        else: res.append(R[j]); j += 1
    return res + L[i:] + R[j:]

t0 = time.time(); merge_sort(sample); t1 = time.time()
print(f'Merge Sort  (n=100) → {(t1-t0)*1000:.4f} ms | Complexity: O(n log n)')
```

```python
# Binary Search — O(log n)
def binary_search(arr, target):
    lo, hi = 0, len(arr)-1
    while lo <= hi:
        mid = (lo+hi)//2
        if arr[mid] == target: return mid
        elif arr[mid] < target: lo = mid+1
        else: hi = mid-1
    return -1

sp = sorted(df['Close'].tolist())
target = sp[len(sp)//2]
t0 = time.time(); idx = binary_search(sp, target); t1 = time.time()
print(f'Binary Search       → Target=${target:.2f} | idx={idx} | {(t1-t0)*1000:.6f} ms | O(log n)')
```

```python
# Complexity Comparison Chart
n = np.arange(1, 101)
plt.figure(figsize=(10, 5))
plt.plot(n, np.log2(n),   label='O(log n)',   color='purple', lw=2)
plt.plot(n, n,            label='O(n)',        color='green',  lw=2)
plt.plot(n, n*np.log2(n), label='O(n log n)', color='blue',   lw=2)
plt.plot(n, n**2,         label='O(n²)',       color='red',    lw=2)
plt.ylim(0, 500)
plt.title('Time Complexity Comparison\nM. Umer Abdullah | SAP: 80393 | Sir Krar Haider',
          fontsize=13, fontweight='bold')
plt.xlabel('Input Size (n)'); plt.ylabel('Operations')
plt.legend(fontsize=11); plt.tight_layout(); plt.show()
```
